```python
## German lexicon used later for correction
from pathlib import Path

print("CWD:", Path.cwd())

dic_path = Path("de.dic")     
out_path = Path("../src/words_de.txt")
```

```python
lines = dic_path.read_text(encoding="utf-8", errors="ignore").splitlines()
```

```python
words = []
for line in lines:
    line = line.strip()
    if not line:
        continue
    # Skip header / license text 
    if " " in line:
        continue


    if line.isdigit():
        continue
    # Strip flags: "Haus/AB" -> "Haus"
    w = line.split("/", 1)[0].strip()
    if w:
        words.append(w)
```

```python
seen = set()
dedup = []
for w in words:
    if w not in seen:
        seen.add(w)
        dedup.append(w)

out_path.write_text("\n".join(dedup) + "\n", encoding="utf-8")

print(f"Wrote {len(dedup):,} words to: {out_path.resolve()}")
print("First 20:", dedup[:20])
```
